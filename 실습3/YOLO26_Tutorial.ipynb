# 5. Python 사용법

YOLO26은 Python에서 YOLO를 더욱 자연스럽게 사용할 수 있도록 Python 중심의 설계 원칙에 따라 재설계되었습니다. YOLO26 모델은 학습된 체크포인트에서 불러오거나 처음부터 생성할 수 있습니다. 그다음 메서드를 사용하여 모델을 학습, 검증, 예측 및 내보내기할 수 있습니다. 자세한 Python 사용 예제는 [YOLO26 Python 문서](https://docs.ultralytics.com/usage/python)를 참고하세요.


```
from ultralytics import YOLO

# 모델 불러오기
model = YOLO('yolo26n.yaml')  # 새로운 모델을 처음부터 생성
model = YOLO('yolo26n.pt')  # 사전 학습된 모델 불러오기 (학습 시 권장)

# 모델 사용하기
results = model.train(data='coco8.yaml', epochs=3)  # 모델 학습
results = model.val()  # 검증 데이터셋에서 모델 성능 평가
results = model('https://ultralytics.com/images/bus.jpg')  # 이미지에 대한 예측 수행
results = model.export(format='onnx')  # 모델을 ONNX 형식으로 내보내기
```

## 1. 객체 검출

YOLO26의 _객체 검출_ 모델에는 접미사가 없으며, 기본 YOLO26 모델에 해당합니다. 예를 들어 `yolo26n.pt`가 있으며, COCO로 사전 학습되어 있습니다. 자세한 내용은 [객체 검출 문서](https://docs.ultralytics.com/tasks/detect)를 참고하세요.


```
# YOLO26n을 불러와 COCO8로 3에포크 동안 학습한 뒤, 이미지에 대한 예측 수행
from ultralytics import YOLO

model = YOLO('yolo26n.pt')  # 사전 학습된 YOLO 객체 검출 모델 불러오기
model.train(data='coco8.yaml', epochs=3)  # 모델 학습
model('https://ultralytics.com/images/bus.jpg')  # 이미지에 대한 예측 수행
```

## 2. 인스턴스 분할

YOLO26의 _인스턴스 분할_ 모델은 `-seg` 접미사를 사용합니다. 예를 들어 `yolo26n-seg.pt`가 있으며, COCO로 사전 학습되어 있습니다. 자세한 내용은 [인스턴스 분할 문서](https://docs.ultralytics.com/tasks/segment)를 참고하세요.


```
# YOLO26n-seg를 불러와 COCO8-seg로 3에포크 동안 학습한 뒤, 이미지에 대한 예측 수행
from ultralytics import YOLO

model = YOLO('yolo26n-seg.pt')  # 사전 학습된 YOLO 인스턴스 분할 모델 불러오기
model.train(data='coco8-seg.yaml', epochs=3)  # 모델 학습
model('https://ultralytics.com/images/bus.jpg')  # 이미지에 대한 예측 수행
```
